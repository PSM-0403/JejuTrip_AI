```python
import pandas as pd

# 1. 파일 로드 (파일명이 본인 컴퓨터에 있는 것과 같은지 확인!)
origin_df = pd.read_csv('jeju_최종_리스트.csv',encoding='cp949') # 위경도, 주소, URL 등 포함
crawled_df = pd.read_csv('jeju_final_combined.csv') # 평점, 리뷰 개수, 리뷰 텍스트 포함

# 2. 데이터 병합 (이름 'name'을 기준으로 합치기)
# 'how=left'는 원본 리스트 1,234개를 모두 유지하겠다는 뜻입니다.
# crawled_df에도 category가 있으므로, 병합 전 중복 방지를 위해 삭제 후 합칩니다.
final_df = pd.merge(origin_df, crawled_df.drop(columns=['category']), on='name', how='left')

# 3. 데이터 검증 (잘 합쳐졌는지 확인)
print("--- 병합 결과 확인 ---")
print(f"전체 데이터 개수: {len(final_df)}개") # 1,234개가 나와야 함
print(f"리뷰가 있는 데이터 개수: {final_df['reviews_text'].notna().sum()}개")

# 4. 최종 마스터 파일 저장
# 한글 깨짐 방지를 위해 'utf-8-sig' 인코딩을 사용합니다.
final_df.to_csv('jeju_travel_master.csv', index=False, encoding='utf-8-sig')

print("\n'jeju_travel_master.csv' 파일로 합치기 성공!")
```

```python
import pandas as pd

# 1. 마스터 데이터 로드
df = pd.read_csv('jeju_travel_master.csv')

# 2. 필터링: kakao_cnt가 10 이상인 행만 유지
# (10개 미만, 즉 0~9개인 행은 삭제됩니다.)
initial_count = len(df)
df_filtered = df[df['kakao_cnt'] >= 10].copy()
final_count = len(df_filtered)

# 3. 삭제된 개수 확인
dropped_count = initial_count - final_count
print(f"원래 데이터: {initial_count}개")
print(f"삭제된 데이터 (리뷰 10개 미만): {dropped_count}개")
print(f"최종 남은 데이터: {final_count}개")

# 4. 진짜_진짜_최종_마스터 저장
df_filtered.to_csv('jeju_travel_final_refined.csv', index=False, encoding='utf-8-sig')

print("\n'jeju_travel_final_refined.csv' 저장 완료! 이제 깨끗한 데이터만 남았습니다.")
```
