```python
import pandas as pd

# 1. 크롤링 결과 파일 로드
df = pd.read_csv('jeju_crawling.csv')

# 2. 실패 조건 설정
# 조건: 카카오 리뷰 개수는 0보다 큰데, 수집된 리뷰 텍스트가 결측치(NaN)인 경우
fail_df = df[(df['kakao_cnt'] > 0) & (df['reviews_text'].isna())]

# 3. 실패 리스트 확인 (상위 20개)
print(f"총 실패 항목 수: {len(fail_df)}개")
print(fail_df[['name', 'kakao_cnt', 'rating']].head(5))

# 4. 실패 리스트만 따로 저장 (나중에 재크롤링용)
fail_df.to_csv('jeju_fail_list.csv', index=False, encoding='utf-8-sig')

print("\n'jeju_fail_list.csv'로 저장 완료!")
```
