```python
import pandas as pd

# 1. 두 파일 로드
df1 = pd.read_csv('jeju_crawling.csv')
df2 = pd.read_csv('jeju_crawling2.csv')

# 2. 재크롤링 결과(df2)에서 성공한 리뷰들만 추출하여 맵핑용 딕셔너리 생성
update_map = df2[df2['reviews_text'].notna()].set_index('name')['reviews_text'].to_dict()

# 3. 기존 데이터(df1)에서 비어있는 리뷰 칸을 재크롤링 결과로 채워넣기
def update_reviews(row):
    # 리뷰가 비어있는 경우에만 df2에서 가져온 값을 넣습니다.
    if pd.isna(row['reviews_text']) or str(row['reviews_text']).strip() == "":
        return update_map.get(row['name'], row['reviews_text'])
    return row['reviews_text']

df1['reviews_text'] = df1.apply(update_reviews, axis=1)

# 4. 최종 파일 저장
df1.to_csv('jeju_final_combined.csv', index=False, encoding='utf-8-sig')

print("데이터 합치기 완료! 'jeju_final_combined.csv' 파일을 확인하세요.")
```
